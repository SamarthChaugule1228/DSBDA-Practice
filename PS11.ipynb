```python
import pandas as pd 
import numpy as np 
```

```python
df = pd.read_csv("./DSBDALExam/Movie/movies_metadata.csv")
```

```python
df.head(-2)
```

```python
df.columns
```

```python
# a) Create data subsets for different languages(Original Language).
english = df[df['original_language'] =='en']
french = df[df['original_language'] == 'fa']
print("English Language Data set :\n ")
print(english.head())
print("French Language DataSet : \n ")
print(french.head())
```

```python
# b) n. Merge two subsets 
merged_data = pd.concat([english , french])
print("Merged Data : \n")
print(merged_data.head(-4))
```

```python
#  c) o. Sort Data using customer ratings.

sorted_data = df.sort_values(by="vote_average" , ascending=False )
print("Sorted Data : \n ")
print(sorted_data.head(-4))
```

```python
# d)  p. Transposing Data
transposed_data = df.head().T
print("Transposed Data : \n")
print(transposed_data)
```

```python
# e) q. Melting Data to long format 
# df['title'].head(
melting_data = pd.melt(
    df ,
    id_vars='title',
    value_vars=['popularity' , 'vote_count' , 'vote_average'],
    var_name='Metrics',
    value_name='Value'
)
print("Melting Data : \n")
print(melting_data.head(-4))
```

```python
# f ) r. Casting data to wide format 
wide_data = melting_data.pivot_table(
    index='title',
    columns= 'Metrics',
    values= 'Value',
    aggfunc='first'
)
print( "Casting Data : \n")
print(wide_data.head())
```
