```python
import pandas as pd
```

```python
columns=[
    'Class',
     'AGE',
     'SEX',
     'STEROID',
    'ANTIVIRALS',
    'FATIGUE',
    'MALAISE',
    'ANOREXIA',
     'LIVER-BIG',
    'LIVER-FIRM',
    'SPLEEN-PALPABLE',
    'SPIDERS',
    'ASCITES',
    'VARICES',
    'BILIRUBIN',
    'ALK PHOSPHATE',
    'SGOT', 
    'ALBUMIN',
    'PROTIME',
    'HISTOLOGY'
]
```

```python
df=  pd.read_csv("./DSBDALExam/Hepatitis/hepatitis.csv" , names=columns)
df.head()
```

```python
# a. Create data subsets for different sex.
male_data = df[df['SEX'] == 1]
print("Male Data :\n" )
# print(male_data.head())
female_data = df[df['SEX'] == 2]
print("female Data : \n")
print( female_data.head())
```

```python
# b. Merge two subsets 
merged_data = pd.concat([male_data , female_data])
print("Merged Data :\n ")
print(merged_data.head(-4))
```

```python
# c. Sort Data using age, SGOT, PROTIME. 
sorted_data = df.sort_values( by=['AGE' ,'SGOT' ,'PROTIME'] , ascending=[False , False , False ])
print("Sorted Data : \n")
print(sorted_data.head())
```

```python
# d. Transposing Data 
transposed_data = df.head().T
print("Transposed Data : \n")
print(transposed_data.head())
```

```python
# E. e. Melting Data to long format 
melting_data = pd.melt(
    df, 
    id_vars='SEX',
    value_vars=['AGE' , 'SGOT' , 'PROTIME'],
    var_name='Metrics' , 
    value_name='Value'
)
print("Melting Data : \n")
print(melting_data.head(-4))
```

```python
# f. Casting data to wide format 
wide_data = melting_data.pivot_table(
    index="SEX",
    columns="Metrics",
    values='Value',
    aggfunc='first'
)
print("Wide Data : \n" )
print(wide_data.head(-4))
```
