## Create a review scrapper for https://store.steampowered.com/appreviews/730?json=1  website to fetch real time comments, reviews, ratings, comment tags, customer name using Python. 


```python
# import the libraries 
import pandas as pd 
import requests
```

import requests
import pandas as pd

# Steam Review API
url = "https://store.steampowered.com/appreviews/730?json=1"

# Get data
data = requests.get(url).json()

reviews = []

# Extract reviews
for r in data['reviews']:

    reviews.append({
        "Steam_ID": r['author']['steamid'],
        "Comment": r['review'],
        "Rating": "Positive" if r['voted_up'] else "Negative",
        "Helpful_Votes": r['votes_up']
    })

# Convert to table
df = pd.DataFrame(reviews)

# Print reviews
print(df)

# Save CSV file
df.to_csv("steam_reviews.csv", index=False)

print("Reviews saved successfully!")
