```python
import pandas as pd 
import numpy as np
```

```python
#Load the DataSet 
df = pd.read_csv("./DSBDALExam/dataset_Facebook.csv" , sep=';')
df.head(5)
```

```python
#columns 
df.columns
```

```python
photo_posts = df[df['Type'] == 'Photo']
print("photo post Subset : \n")
print(photo_posts.head(1))
```

```python
status_post = df[df['Type'] == 'Status']
print( status_post.head())
```

```python
# b)  Merge two subsets 
merged_data = pd.concat([status_post , photo_posts] ,ignore_index=True)
print("merged Subset : \n ")
print( merged_data.head())
```

```python
# c) Sort Data on Page total likes
sorted_data = df.sort_values(by="Page total likes" , ascending=True)
print("After Sorting :\n ")
print(sorted_data.head())
```

```python
# d)  Transposing Data 
transpose_data = df.head().T
print("Transposed Data : \n ")
print(transpose_data.head())
```

```python
# e) Melting Data to long format 
melted_data = pd.melt(
    df ,
    id_vars=['Type'],
    value_vars=['like' , 'share' , 'comment'],
    var_name='Metrics',
    value_name='Count'
)
melted_data.head()
```

```python
# f) Casting data to wide format (long to wide )
wide_data = melted_data.pivot_table(
    index='Type' ,
    columns='Metrics' ,
    values='Count' ,
    aggfunc='mean'
)
print("Long to Wide Data : \n")
print( wide_data.head())
```
