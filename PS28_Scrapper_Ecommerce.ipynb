## Create a review scrapper for https://webscraper.io/test-sites/e-commerce/static website to fetch real time comments, reviews, ratings, comment tags, customer name using Python. 
- code
 import requests
from bs4 import BeautifulSoup
import pandas as pd

# Website URL
url = "https://webscraper.io/test-sites/e-commerce/static"

# Fetch webpage
response = requests.get(url)

# Parse HTML
soup = BeautifulSoup(response.text, "html.parser")

data = []

# Find products
products = soup.find_all("div", class_="thumbnail")

for p in products:

    name = p.find("a", class_="title").text.strip()

    price = p.find("h4", class_="price").text.strip()

    desc = p.find("p", class_="description").text.strip()

    rating = len(p.find_all("span", class_="ws-icon-star"))

    review = p.find("p", class_="review-count").text.strip()

    data.append({
        "Product Name": name,
        "Price": price,
        "Description": desc,
        "Rating": rating,
        "Reviews": review,
        "Tag": "Good" if rating >= 4 else "Average"
    })

# Create DataFrame
df = pd.DataFrame(data)

# Print output
print(df)

# Save CSV
df.to_csv("reviews.csv", index=False)

print("CSV file saved")
