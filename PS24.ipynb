```python
import pandas as pd
```

```python
df = pd.read_csv("./DSBDALExam/forestfires/forestfires.csv")
df.head()
```

```python
def classify_area(area):
    if(area == 0):
        return "NotAffected"
    elif(area<=50):
        return "PartiallyAffected"
    else:
        return "MostlyAffected"
df['Affected_class'] = df['area'].apply(classify_area)
print(df.head(-4))
```

```python
# a)  Create data subsets by making classes for amount of region affected.(e.g. NotAffected, Partially affected, Mostlyaffected).
not_affected = df[df['Affected_class'] == 'NotAffected']
partially_affected = df[df['Affected_class']=='PartiallyAffected']
mostly_affected = df[df['Affected_class'] =='MostlyAffected']
print("Not Affected : \n ")
print( not_affected.head())
print( "Partically Affected : \n ")
print( partially_affected.head())
print("Mostly Affected : \n ")
print(mostly_affected.head())
```

```python
# b. Merge two subsets 
merged_data = pd.concat( [not_affected , partially_affected])
print("Merged Data : \n ")
print( merged_data.head(-4))
```

```python
# c. Sort Data using Temperature, wind and area. 

sorted_data = df.sort_values( by=['temp' , 'wind' , 'area'] , ascending=[False , False , False ])
print(sorted_data[['temp' , 'wind' , 'area']].head())
```

```python
# D) TAnsposing the Data
transposed_data = df.head().T
print("Transposed Data : ")
print(transposed_data.head())
```

```python
# E) Melting Data to long format
melting_data = pd.melt(
    df, 
    id_vars='month',
    value_vars=['temp', 'wind' , 'RH'],
    var_name='Metrics',
    value_name='Value'
)
print("Melting Data :\n ")
print(melting_data.head())
```

```python
# f )  Casting data to wide format 
wide_data = melting_data.pivot_table(
    index='month',
    columns='Metrics',
    values ='Value',
    aggfunc='mean'
)
print("Wide Data : \n")
print(wide_data)

```
