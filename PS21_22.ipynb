## 22 Visualize the census bureau databset(Adult data sets)by plotting the following graphs using Python. (Define objective for every graph) 
    a. Histograms 
    b. Pie Charts 
    c. Box Plots 
    d. Scatter Plots 
    e. Add boxplots to a scatterplot

```python
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
columns = [
    "age",
    "workclass",
    "fnlwgt",
    "education",
    "education_num",
    "marital_status",
    "occupation",
    "relationship",
    "race",
    "sex",
    "capital_gain",
    "capital_loss",
    "hours_per_week",
    "native_country",
    "income"
]
```

```python
df = pd.read_csv("./DSBDALExam/Adult/adult.csv" , names=columns)
df.head()
```

```python

# =========================================================
# A) HISTOGRAM
# Objective:
# To understand the distribution of age of people.
# =========================================================

plt.figure(figsize=(7,5))
plt.hist(df['age'], bins=10 , color='green' , edgecolor='black')
plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

```

```python

# =========================================================
# B) DOT PLOT
# Objective:
# To visualize individual working hours of people.
# =========================================================

plt.figure(figsize=(10,4))
plt.plot(df['hours_per_week'], 'o')
plt.title("Dot Plot of Hours Per Week")
plt.xlabel("Person Index")
plt.ylabel("Hours Per Week")
plt.show()
```

```python

# =========================================================
# C) BAR PLOT
# Objective:
# To compare number of males and females.
# =========================================================

gender_count = df['sex'].value_counts()

plt.figure(figsize=(6,5))
plt.bar(gender_count.index, gender_count.values)
plt.title("Bar Plot of Gender Count")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python

# =========================================================
# D) LINE CHART
# Objective:
# To observe variation in education number.
# =========================================================

plt.figure(figsize=(10,5))
plt.plot(df['age'])
plt.title("Line Chart of Education Number")
plt.xlabel("Person Index")
plt.ylabel("Education Number")
plt.show()
```

```python
# =========================================================
# E) PIE CHART
# Objective:
# To show percentage distribution of income classes.
# =========================================================

income_count = df['income'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(income_count.values,
        labels=income_count.index,
        autopct='%1.1f%%')

plt.title("Pie Chart of Income Distribution")
plt.show()
```

```python
# =========================================================
# F) BOX PLOT
# Objective:
# To detect spread and outliers in age values.
# =========================================================

plt.figure(figsize=(6,5))
plt.boxplot(df['age'])
plt.title("Box Plot of Age")
plt.ylabel("Age")
plt.show()

```

```python
# =========================================================
# G) SCATTER PLOT
# Objective:
# To study relationship between age and hours-per-week.
# =========================================================

plt.figure(figsize=(7,5))
plt.scatter(df['age'], df['hours_per_week'])
plt.title("Scatter Plot of Age vs Hours Per Week")
plt.xlabel("Age")
plt.ylabel("Hours Per Week")
plt.show()
```

```python
# =========================================================
# H) HISTOGRAM + BOX PLOT
# Objective:
# To analyze distribution and outliers in capital-gain.
# =========================================================

plt.figure(figsize=(6,5))
plt.boxplot(df['capital_gain'])
plt.title("Box Plot of Capital Gain")
plt.ylabel("Capital Gain")
plt.show()

plt.figure(figsize=(7,5))
plt.hist(df['capital_gain'], bins=10)
plt.title("Histogram of Capital Gain")
plt.xlabel("Capital Gain")
plt.ylabel("Frequency")
plt.show()
```

```python
# =========================================================
# I) SCATTER PLOT + BOX PLOT
# Objective:
# To identify relationship and spread in age and fnlwgt.
# =========================================================

plt.figure(figsize=(7,5))
plt.scatter(df['age'], df['fnlwgt'])
plt.title("Scatter Plot of Age vs fnlwgt")
plt.xlabel("Age")
plt.ylabel("fnlwgt")
plt.show()

plt.figure(figsize=(6,5))
plt.boxplot(df['fnlwgt'])
plt.title("Box Plot of fnlwgt")
plt.ylabel("fnlwgt")
plt.show()
```
