## 18) Visualize the Heart disease dataset by plotting the following graphs using Python. (Define objective for every graph) 
    a. Histograms 
    b. Pie Charts 
    c. Box Plots 
    d. Scatter Plots 
    e. Add boxplots to a scatterplot 


```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt 
```

```python
columns = [
    "age", "sex", "cp", "trestbps", "chol", "fbs",
    "restecg", "thalach", "exang", "oldpeak",
    "slope", "ca", "thal", "num"
]
```

```python
df = pd.read_csv("./DSBDALExam/HeartDisease/Cleavland.csv" , header=None  , names=columns)
df.head()
```

```python
df.columns
```

```python
# =========================================================
# A) HISTOGRAM
# Objective:
# To understand the distribution of age of patients.
# =========================================================
plt.figure(figsize=(9, 10 ))
plt.hist( df['age'] , bins=10 , color="skyblue" , edgecolor="pink")
plt.title("Distrubution of age of patients ")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
# =========================================================
# B) DOT PLOT
# Objective:
# To visualize individual cholesterol values.
# =========================================================
plt.figure(figsize=(10, 9 ))
plt.plot( df['chol'] ,'o' , color="red" )
plt.title("Dot plot of cholesterol Value" )
plt.xlabel("Patient Index")
plt.ylabel( "Cholesteroel of Patients ")
plt.show()
```

```python
# =========================================================
# C) BAR PLOT
# Objective:
# To compare number of male and female patients.
# sex: 1 = Male, 0 = Female
# =========================================================
sex_counts = df['sex'].value_counts()

plt.figure( figsize=( 9 , 10 ))
plt.bar( ['Male' , 'Female'] , sex_counts.values , edgecolor= "black" , color='green')
plt.title( "DBar Plot of Gender Compare ")
plt.xlabel("Gender" )
plt.ylabel("Count")
plt.show()
```

```python
# =========================================================
# D) LINE CHART
# Objective:
# To observe variation in resting blood pressure.
# =========================================================
plt.figure(figsize=(10, 9 ))
plt.plot(df['trestbps'] , color = 'red' )
plt.title("line Chart for the resting blood pressure ")
plt.xlabel("patient Index")
plt.ylabel("resting blood pressure ")
plt.show()
```

```python
# =========================================================
# E) PIE CHART
# Objective:
# To show percentage distribution of chest pain types.
# pie chart for the chest pain types 
# =========================================================
cp_count = df['cp'].value_counts()

plt.figure(figsize=(6, 6 ))
plt.pie( cp_count.values , labels= cp_count.index  , startangle=90 , shadow=True )
plt.title("Chest Pain Types Pie Graph ")
plt.show()
```

```python
# =========================================================
# F) BOX PLOT
# Objective:
# To detect spread and outliers in cholesterol values.
# ============+++++++++++++++++++++++===========================
plt.figure(figsize=( 6,6 ) )
plt.boxplot( df['chol'] )
plt.title("Box Plot of Cholestrerol")
plt.ylabel("cholesterol")
plt.show()
```

```python
# =========================================================
# G) SCATTER PLOT
# Objective:
# To study relationship between age and cholesterol.
# =========================================================

plt.figure(figsize=(7,7))
plt.scatter(df['age'] , df['chol'])
plt.title("Age VS Cholesterol")
plt.xlabel("Age")
plt.ylabel("Cholesterol")
plt.show()
```

```python

# =========================================================
# H) HISTOGRAM + BOX PLOT
# Objective:
# To analyze distribution and outliers in oldpeak values.
# =========================================================

plt.figure(figsize=(6,5))
plt.boxplot(df['oldpeak'])
plt.title("Box Plot of Oldpeak")
plt.ylabel("Oldpeak")
plt.show()

plt.figure(figsize=(7,5))
plt.hist(df['oldpeak'], bins=10)
plt.title("Histogram of Oldpeak")
plt.xlabel("Oldpeak")
plt.ylabel("Frequency")
plt.show()
```

```python
# =========================================================
# I) SCATTER PLOT + BOX PLOT
# Objective:
# To identify relationship and spread in age and trestbps.
# =========================================================

plt.figure(figsize=(7,5))
plt.scatter(df['age'], df['trestbps'])
plt.title("Scatter Plot of Age vs Resting BP")
plt.xlabel("Age")
plt.ylabel("Resting BP")
plt.show()

plt.figure(figsize=(6,5))
plt.boxplot(df['trestbps'])
plt.title("Box Plot of Resting BP")
plt.ylabel("Resting BP")
plt.show()
```
