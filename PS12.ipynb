```python
import pandas as pd 
import numpy as np 
```

```python
columns=[
    'age',
    'workclass',
    'fnlwgt',
    'education',
    'education-num',
    'marital-status',
    'occupation',
    'relationship',
    'race',
    'sex',
    'capital-gain',
    'capital-loss',
    'hours-per-week',
    'native-country',
    'income'
]
```

```python
df = pd.read_csv('./DSBDALExam/Adult/adult.csv' , names=columns)
df.head()
```

```python
# a)s. Create data subsets for different Country, Sex, race. 
country = df[df['native-country']==' Cuba']
sex_data = df[df['sex'] == ' Male']
race_data = df[df['race'] == ' Black']
print("Country Subset : \n ")
print(country.head())
print("Sex Subset : \n ")
print(sex_data.head())
print("Race Subset : \n ")
print(race_data.head())
```

```python
# b)  t. Merge two subsets 
merged_data = pd.concat([country , race_data], ignore_index=True)
print("Merged Data : ")
print(merged_data.head())
```

```python
#c ) u. Sort Data using customer ratings.(capital Gain)
sorted_data = df.sort_values( by='capital-gain' , ascending=False)
print("Sorted Data :\n ")
print(sorted_data.head())
```

```python
# d) Transposing Data 
transpose_data = df.head().T
print("Transposed Data : \n")
print(transpose_data.head())
```

```python
# e) Melting Data to long format
melting_data = pd.melt(
    df , 
    id_vars='sex',
    value_vars=['education-num' , 'age' , 'hours-per-week'],
    var_name='Metrics',
    value_name='Value'
)
print("Melting data to Long Format : ")
print(melting_data.head(-4))
```

```python
# f)  x. Casting data to wide format 
wide_data = melting_data.pivot_table(
    index='sex',
    columns='Metrics',
    values = 'Value',
    aggfunc='mean'
)
print("Wide Data : \n")
print(wide_data)
```
