```python
import pandas as pd 
import numpy as np 
```

```python
# Rename the Columns 
# Attribute Information:
#    1. sepal length in cm
#    2. sepal width in cm
#    3. petal length in cm
#    4. petal width in cm
#    5. class: Species
columns=[
    'sepalLength',
    'sepalWidth' ,
    'petalLength',
    'petalWidth',
    'Species'
]

```

```python
df = pd.read_csv("./DSBDALExam/Iris/Iris.csv"  ,names=columns )
df.info()
```

```python
df.head()
```

```python
df.columns
```

```python
# G) Create data subsets for different species
setosa = df[df['Species'] == 'Iris-setosa']
versicolor =df[df['Species'] =='Iris-versicolor']
virginica = df[df['Species'] == 'Iris-virginica']
```

```python
#display subsets --> filter the Rows 
print("satosa :\n ")
print(setosa.head())
print(" \n Versicolor : \n ")
print(versicolor.head())
print("\n Virginica : \n")
print(virginica.head())
```

```python
# H) Merge two subsets 
merged_data = pd.concat( [ setosa  , versicolor ] )
print("Merged Data : \n ")
print(merged_data.head(-5))
```

```python
# I) Sort the data by petal length 
sorted_data = df.sort_values(by='petalLength' , ascending=True)
print("Sorted Values : \n")
print(sorted_data.head())
```

```python
# j. Transposing Data
transposed_data = df.head().T
print("Transposed Data : \n")
print(transposed_data)
```

```python
# k. Melting Data to long format 
melting_data = pd.melt(
    df , 
    id_vars='Species',
    value_vars=['sepalLength' , 'petalLength'],
    var_name='Measurement',
    value_name='Value'
)
```

```python
print("Melted Long Data : \n ")
print(melting_data.head(-5))
```

```python
# l. Casting data to wide format 

wide_data = melting_data.pivot_table(
    index='Species',
    columns='Measurement',
    values='Value',
    aggfunc='mean'
)
print("Wide Data : \n ")
print(wide_data.head())
```
